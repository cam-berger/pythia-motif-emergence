# Per-size motif co-evolution movies

Side-by-side animations of the three attention-head motifs (induction, successor, S-inhibition) evolving across the §H2-1 40-checkpoint grid, one animation per Pythia size. Each panel shows the (layer, head) score grid for that motif, with a shared frame index across the 3 panels so motif co-emergence is visually comparable.

**Per-motif color scales (separate colorbars per panel):**
- Induction: prefix-matching score, range [0, 1]
- Successor: §SU-1b lift score (real_DLA − null_DLA), range varies by size
- S-inhibition: §S-1 path-patching Δ_h scalar, range varies by size

**2.8B caveat (§H4-7-supersede):** the 2.8B S-inhibition sweep was halted at 8/40 cells. The 2.8B animation therefore has only 2 panels (induction + successor); the S-inhibition panel is omitted. The 8 partial S-inhibition cells at 2.8B are preserved as `data/exploration/phase4_2_8b_s_inhibition_per_prompt/` but are not animated here.

## Setup

```python
import os
os.environ.setdefault('PYTORCH_ENABLE_MPS_FALLBACK', '1')
os.environ.setdefault('HF_HUB_OFFLINE', '1')

import sys
from pathlib import Path
REPO = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(REPO) not in sys.path:
    sys.path.insert(0, str(REPO))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML, display
from notebooks._lib.sweep_io import read_long

INDUCTION_THRESHOLD = 0.3
TAU_LIFT = 0.13496
TAU_STRICT = 0.0372

def sweep_path(motif: str, size: str) -> Path:
    if size == '1b':
        return REPO / 'data' / 'exploration' / f'phase3_1b_{motif}_sweep.parquet'
    if size == '2.8b':
        return REPO / 'data' / 'exploration' / f'phase4_2_8b_{motif}_sweep.parquet'
    return REPO / 'data' / 'exploration' / f'phase2_{motif}_sweep.parquet'

STEPS = [0, 1, 2, 4, 8, 16, 32, 64, 128, 256, 512,
         1000, 2000, 3000, 4000, 5000, 6000, 7000, 8000, 9000,
         10000, 11000, 12000, 13000, 14000, 15000, 16000, 17000,
         20000, 24000, 29000, 35000, 41000, 49000, 59000, 70000,
         84000, 100000, 120000, 143000]

FIG_DIR = REPO / 'notebooks' / 'figures'
FIG_DIR.mkdir(parents=True, exist_ok=True)
```

## Load all motif sweep parquets (5 sizes × 3 motifs, dedup phase2)

```python
# phase2_*_sweep.parquet contains all 3 phase2 sizes; load once per motif then filter.
df_ind = pd.concat([
    read_long(REPO / 'data' / 'exploration' / 'phase2_induction_sweep.parquet'),
    read_long(REPO / 'data' / 'exploration' / 'phase3_1b_induction_sweep.parquet'),
    read_long(REPO / 'data' / 'exploration' / 'phase4_2_8b_induction_sweep.parquet'),
]).reset_index(drop=True)
df_suc = pd.concat([
    read_long(REPO / 'data' / 'exploration' / 'phase2_successor_sweep.parquet'),
    read_long(REPO / 'data' / 'exploration' / 'phase3_1b_successor_sweep.parquet'),
    read_long(REPO / 'data' / 'exploration' / 'phase4_2_8b_successor_sweep.parquet'),
]).reset_index(drop=True)
df_si = pd.concat([
    read_long(REPO / 'data' / 'exploration' / 'phase2_s_inhibition_sweep.parquet'),
    read_long(REPO / 'data' / 'exploration' / 'phase3_1b_s_inhibition_sweep.parquet'),
]).reset_index(drop=True)
# 2.8b s_inhibition omitted (§H4-7-supersede).
MOTIF_DF = {'induction': df_ind, 'successor': df_suc, 's_inhibition': df_si}
MOTIF_LABEL = {'induction': 'Induction (prefix-match)',
               'successor': 'Successor (lift_dla)',
               's_inhibition': 'S-inhibition (Δ_h)'}
MOTIF_THRESHOLD = {'induction': INDUCTION_THRESHOLD,
                   'successor': TAU_LIFT,
                   's_inhibition': TAU_STRICT}
for motif, df in MOTIF_DF.items():
    print(f'{motif}: {len(df):,} rows, sizes={sorted(df["size"].unique().tolist())}')
```

## Helper: build heatmap-frames for one (size, motif)

Returns a list of (n_layers, n_heads) arrays — one per step in STEPS — and the (vmin, vmax) for that panel's color scale.

```python
def build_frames(size: str, motif: str):
    df = MOTIF_DF[motif]
    sub = df[df['size'] == size]
    if sub.empty:
        return None, None, None  # signals 'skip this panel'
    n_layers = int(sub['layer'].max() + 1)
    n_heads = int(sub['head'].max() + 1)
    frames = []
    for step in STEPS:
        cell = sub[sub['step'] == step]
        grid = np.zeros((n_layers, n_heads))
        for _, r in cell.iterrows():
            grid[int(r['layer']), int(r['head'])] = r['score']
        frames.append(grid)
    all_vals = np.concatenate([f.flatten() for f in frames])
    return frames, float(all_vals.min()), float(all_vals.max())
```

## Render per-size animations

5 sizes × 1 animation each. Each animation is built with `matplotlib.animation.FuncAnimation`, saved as a GIF at 3 fps (13 s playback for 40 frames), and embedded inline as an HTML5 player.

```python
SIZES = ['70m', '160m', '410m', '1b', '2.8b']
MOTIFS = ['induction', 'successor', 's_inhibition']

for size in SIZES:
    # Determine which motifs have data for this size.
    motifs_for_size = []
    motif_data = {}
    for motif in MOTIFS:
        frames, vmin, vmax = build_frames(size, motif)
        if frames is None:
            continue
        motifs_for_size.append(motif)
        motif_data[motif] = (frames, vmin, vmax)
    n_panels = len(motifs_for_size)
    if n_panels == 0:
        print(f'{size}: no motif data — skipping')
        continue
    fig, axes = plt.subplots(1, n_panels, figsize=(5.5 * n_panels, 5))
    if n_panels == 1:
        axes = [axes]
    ims = []
    for ax, motif in zip(axes, motifs_for_size):
        frames, vmin, vmax = motif_data[motif]
        cmap = 'viridis' if vmin >= 0 else 'RdBu_r'
        if cmap == 'RdBu_r':
            limit = max(abs(vmin), abs(vmax))
            im = ax.imshow(frames[0], vmin=-limit, vmax=limit, cmap=cmap, aspect='auto')
        else:
            im = ax.imshow(frames[0], vmin=vmin, vmax=vmax, cmap=cmap, aspect='auto')
        thr = MOTIF_THRESHOLD[motif]
        ax.set_title(f'{MOTIF_LABEL[motif]}\nthreshold = {thr}', fontsize=10)
        ax.set_xlabel('head')
        ax.set_ylabel('layer')
        plt.colorbar(im, ax=ax, fraction=0.04, pad=0.02)
        ims.append(im)
    title = fig.suptitle(
        f'Pythia-{size} — motif co-evolution at step{STEPS[0]} (frame 1/{len(STEPS)})',
        y=1.0,
    )
    if size == '2.8b' and 's_inhibition' not in motifs_for_size:
        # Annotate the omission per §H4-7-supersede
        fig.text(0.99, 0.01,
                 'S-inhibition omitted: sweep halted at 8/40 cells (§H4-7-supersede)',
                 ha='right', va='bottom', fontsize=8, style='italic', color='gray')

    def make_update(ims_local, motifs_local, motif_data_local, title_local, size_local):
        def _update(frame_idx):
            for im, motif in zip(ims_local, motifs_local):
                im.set_array(motif_data_local[motif][0][frame_idx])
            title_local.set_text(
                f'Pythia-{size_local} — motif co-evolution at step{STEPS[frame_idx]} '
                f'(frame {frame_idx+1}/{len(STEPS)})'
            )
            return ims_local + [title_local]
        return _update

    update = make_update(ims, motifs_for_size, motif_data, title, size)
    ani = animation.FuncAnimation(fig, update, frames=len(STEPS), interval=333, blit=False)
    out_gif = FIG_DIR / f'motif_evolution_{size}.gif'
    ani.save(out_gif, writer=animation.PillowWriter(fps=3))
    print(f'  Pythia-{size}: {n_panels} panels, wrote {out_gif.relative_to(REPO)}')
    plt.close(fig)
    display(HTML(f'<h3>Pythia-{size}</h3>'))
    display(HTML(ani.to_jshtml()))
```
